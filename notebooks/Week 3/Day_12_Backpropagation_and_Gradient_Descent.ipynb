# 1. Value vs Change vs Rate of Change

We need to clearly distinguish three things:

- `z` → current value/output
- `Δz` → how much `z` changed
- `dz/dW` → how quickly `z` changes when `W` changes

### Change

Always calculate:

`Change = New - Old`

Example:

Old `z = 10`

New `z = 14`

Therefore:

`Δz = 14 - 10 = +4`

So `z` increased by 4.

### Rate of Change

Suppose:

`W: 2 → 3`

`z: 7 → 11`

Then:

`ΔW = 3 - 2 = 1`

`Δz = 11 - 7 = 4`

Rate of change:

`Δz / ΔW = 4 / 1 = 4`

# 2. Derivative

A derivative tells us the instantaneous rate of change.

For a neural network:

`dLoss/dW`

means:

> How quickly does Loss change when this weight changes?

The sign is important.

### Positive derivative

If:

`dLoss/dW > 0`

then increasing the weight tends to increase Loss.

Therefore gradient descent will move the weight downward.

### Negative derivative

If:

`dLoss/dW < 0`

then increasing the weight tends to decrease Loss.

Therefore gradient descent will move the weight upward.

### Important

A negative `dLoss/dW` does NOT mean that Loss itself is negative.

It means the RATE OF CHANGE of Loss with respect to W is negative.

# 3. Tiny Neuron

Consider a simple neuron:

`z = xW + b`

Suppose:

`x = 3`

`W = 2`

`b = 1`

Calculate:

`z = (3 × 2) + 1`

`z = 7`

So:

`z = 7`

Here `z` is the current output.

It is NOT the same thing as `dz/dW`.

---

### If W changes

Suppose:

`W: 2 → 3`

New z:

`z = (3 × 3) + 1`

`z = 10`

Therefore:

`Δz = 10 - 7 = 3`

and:

`ΔW = 3 - 2 = 1`

Rate of change:

`Δz / ΔW = 3 / 1 = 3`

# 4. Chain Rule

**ADD TO NOTEBOOK**

Suppose our relationship is:

`W → z → Loss`

We want to know:

> How does changing W affect Loss?

Instead of calculating the entire relationship at once, we break it into smaller relationships:

`W → z`

and:

`z → Loss`

The Chain Rule combines these rates of change:

`dLoss/dW = (dLoss/dz) × (dz/dW)`

---

### Example

Suppose:

`dLoss/dz = -2`

`dz/dW = 4`

Then:

`dLoss/dW = (-2) × 4`

`dLoss/dW = -8`

### Meaning

The negative sign tells us that increasing W tends to decrease Loss around this point.

# 5. Backpropagation

Backpropagation works backward through a neural network and uses the Chain Rule to calculate how each parameter affects Loss.

### Forward

Input

↓

Weights

↓

Prediction

↓

Loss

### Backward

Loss

↓

Derivatives

↓

Gradients

### Important distinction

Backpropagation and Gradient Descent are NOT the same thing.

**Backpropagation:**

Calculates gradients.

**Gradient Descent:**

Uses those gradients to update weights.

# 6. Tiny Backpropagation Example

Given:

`x = 2`

`W = 3`

`b = 1`

`target = 10`

### Step 1 — Forward Pass

`z = xW + b`

`z = (2 × 3) + 1`

`z = 7`

---

### Step 2 — Calculate Loss

We'll use:

`Loss = 1/2 × (z - target)^2`

Therefore:

`Loss = 1/2 × (7 - 10)^2`

`Loss = 4.5`

---

### Step 3 — Calculate dLoss/dz

For this Loss function:

`dLoss/dz = z - target`

Therefore:

`dLoss/dz = 7 - 10`

`dLoss/dz = -3`

---

### Step 4 — Calculate dz/dW

Our neuron is:

`z = xW + b`

Since:

`x = 2`

we get:

`dz/dW = 2`

---

### Step 5 — Chain Rule

`dLoss/dW = (dLoss/dz) × (dz/dW)`

Therefore:

`dLoss/dW = (-3) × 2`

`dLoss/dW = -6`

The negative gradient means increasing W tends to decrease Loss.

# 7. Gradient

For one weight we calculate:

`dLoss/dW`

For many weights, we calculate a derivative for each weight.

Example:

`dLoss/dW1 = 2`

`dLoss/dW2 = -4`

`dLoss/dW3 = 1`

Together:

`Gradient = [2, -4, 1]`

### Remember

Derivative:

> Information about one parameter.

Gradient:

> Collection of derivatives for many parameters.

# 8. Gradient Descent

Gradient Descent uses the gradient to update weights.

The standard update is:

`New Weight = Old Weight - Learning Rate × dLoss/dW`

Example:

`Old Weight = 5`

`dLoss/dW = -4`

`Learning Rate = 0.1`

Calculate the movement:

`0.1 × 4 = 0.4`

The derivative is negative, so the weight moves upward.

Therefore:

`New Weight = 5.4`

Using the full calculation:

`5 - (0.1 × -4)`

`5 - (-0.4)`

`= 5.4`

# 9. Positive vs Negative Gradient

Suppose:

`W1 = 5`

`dLoss/dW1 = +2`

and:

`W2 = 5`

`dLoss/dW2 = -3`

Learning rate:

`0.1`

### W1

Positive gradient means W1 should decrease.

`W1 = 5 - (0.1 × 2)`

`W1 = 4.8`

### W2

Negative gradient means W2 should increase.

`W2 = 5 - (0.1 × -3)`

`W2 = 5.3`

### Rule

Positive gradient → weight decreases.

Negative gradient → weight increases.

# 10. Complete Neural Network Learning Loop

Input

↓

Forward Pass 

↓

Prediction 

↓

Loss 

↓

Backpropagation 

↓

Gradients

↓

Gradient Descent

↓

Updated Weights

↓

Forward Pass Again

↓

Repeat

````python
# ADD TO NOTEBOOK

# Simple neuron
x = 2
w = 3
b = 1
target = 10

# Learning rate
learning_rate = 0.1

# -------------------------
# Forward Pass
# -------------------------

z = x * w + b

# -------------------------
# Loss
# -------------------------

loss = 0.5 * (z - target) ** 2

# -------------------------
# Backpropagation
# -------------------------

d_loss_d_z = z - target

d_z_d_w = x

# Chain Rule
d_loss_d_w = d_loss_d_z * d_z_d_w

# -------------------------
# Gradient Descent
# -------------------------

new_w = w - learning_rate * d_loss_d_w

# -------------------------
# Results
# -------------------------

print("z =", z)
print("Loss =", loss)
print("dLoss/dz =", d_loss_d_z)
print("dz/dW =", d_z_d_w)
print("dLoss/dW =", d_loss_d_w)
print("Old Weight =", w)
print("New Weight =", new_w)
````

## 1. Change in a Value

### Formula

`Δx = New Value - Old Value`

### ELI5

Imagine you have ₹100 and now you have ₹120.

How much did your money change?

`120 - 100 = +20`

So the change is `+20`.

If you had ₹100 and now have ₹80:

`80 - 100 = -20`

The negative sign means the value decreased.

### Example

`Old z = 10`

`New z = 14`

`Δz = 14 - 10 = +4`

---

## 2. Average Rate of Change

### Formula

`Rate of Change = Δy / Δx`

or:

`Rate of Change = (New y - Old y) / (New x - Old x)`

### ELI5

Imagine a car travels 100 km in 2 hours.

Its average speed is:

`100 / 2 = 50 km/hour`

The same idea applies here.

We are asking:

> How much did y change for every 1 unit of change in x?

### Example

`W: 2 → 3`

`z: 7 → 11`

Therefore:

`ΔW = 3 - 2 = 1`

`Δz = 11 - 7 = 4`

Rate:

`Δz / ΔW = 4 / 1 = 4`

# 3. Derivative

### Notation

`dy/dx`

### Meaning

> How quickly does y change when x changes?

### ELI5

Imagine you're driving a car.

Your position tells you **where you are**.

Your speed tells you **how quickly your position is changing**.

Similarly:

`y` → current value

`dy/dx` → how quickly y is changing when x changes

### Neural Network Example

`dLoss/dW`

means:

> How quickly does Loss change when Weight changes?

# 4. Neuron Formula

### Formula

`z = xW + b`

Where:

- `x` = input
- `W` = weight
- `b` = bias
- `z` = neuron output before activation

### ELI5

Think of a neuron like a small calculator:

1. Take the input.
2. Multiply it by its importance/weight.
3. Add the bias.
4. Get the output.

### Example

`x = 3`

`W = 2`

`b = 1`

Therefore:

`z = (3 × 2) + 1`

`z = 7`

---

# 5. Derivative of the Neuron With Respect to Weight

For:

`z = xW + b`

we get:

`dz/dW = x`

### ELI5

If:

`z = x × W + b`

then changing W affects z according to the input x.

For example:

`x = 4`

If W increases by 1, z increases by 4.

Therefore:

`dz/dW = 4`

### Example

`x = 4`

`W: 2 → 3`

Old:

`z = 4×2 + 1 = 9`

New:

`z = 4×3 + 1 = 13`

Change:

`Δz = 13 - 9 = 4`

So:

`dz/dW = 4`

---

# 6. Squared Error Loss

### Formula

`Loss = 1/2 × (Prediction - Target)^2`

For our neuron:

`Loss = 1/2 × (z - target)^2`

### ELI5

Loss is basically a "how wrong was I?" score.

If prediction is very close to the target:

→ small Loss

If prediction is far from the target:

→ large Loss

The square makes the difference positive and makes large mistakes more important.

### Example

Prediction:

`z = 7`

Target:

`10`

Difference:

`7 - 10 = -3`

Square:

`(-3)^2 = 9`

Therefore:

`Loss = 1/2 × 9`

`Loss = 4.5`

---

# 7. Derivative of Loss With Respect to Output

For:

`Loss = 1/2 × (z - target)^2`

the derivative is:

`dLoss/dz = z - target`

### ELI5

This tells us:

> If the output z changes, which direction does Loss want to move?

### Example

`z = 7`

`target = 10`

Therefore:

`dLoss/dz = 7 - 10`

`dLoss/dz = -3`

The negative sign tells us that increasing z would reduce Loss around this point.

---

# 8. Chain Rule

### Formula

`dLoss/dW = (dLoss/dz) × (dz/dW)`

### ELI5

Imagine:

`W → z → Loss`

You want to know:

> How does changing W affect Loss?

But W doesn't directly affect Loss.

It affects z first.

Then z affects Loss.

So we follow the path:

`W → z → Loss`

and multiply the rates along the path.

### Example

`dLoss/dz = -3`

`dz/dW = 2`

Therefore:

`dLoss/dW = (-3) × 2`

`dLoss/dW = -6`

---

# 9. Gradient

For multiple weights:

`Gradient = [dLoss/dW1, dLoss/dW2, dLoss/dW3, ...]`

### ELI5

A derivative tells us what is happening with **one weight**.

A gradient is like a report containing the derivative for **every weight**.

### Example

`dLoss/dW1 = 2`

`dLoss/dW2 = -4`

`dLoss/dW3 = 1`

Therefore:

`Gradient = [2, -4, 1]`

---

# 10. Gradient Descent

### Formula

`New Weight = Old Weight - Learning Rate × dLoss/dW`

### ELI5

Imagine you're standing on a hill and want to walk downhill.

The gradient tells you:

> Which direction is uphill?

So you move in the **opposite direction**.

That's why we subtract the gradient.

The learning rate tells you:

> How big of a step should I take?

### Example

`Old Weight = 5`

`dLoss/dW = -4`

`Learning Rate = 0.1`

Therefore:

`New Weight = 5 - (0.1 × -4)`

`New Weight = 5 - (-0.4)`

`New Weight = 5.4`

---

# 11. Weight Update Amount

### Formula

`Update Amount = Learning Rate × Gradient`

### ELI5

The gradient tells you the strength/direction.

The learning rate controls how much of that gradient you actually use for the step.

### Example

`Learning Rate = 0.1`

`Gradient = 4`

Movement:

`0.1 × 4 = 0.4`

So the weight moves by `0.4`.
