## 1. Forward Pass

A neuron receives inputs and calculates:

z = W · x + b

Then we apply an activation function.

For today's examples, we use ReLU:

a = ReLU(z)

The complete forward path is:

x → W → z → ReLU → a → Loss

### ELI5

Think of a neuron like a small calculator:

Input
 ↓
multiply by weights
 ↓
add bias
 ↓
ReLU
 ↓
output

## 2. Simple Neuron Example

Given:

x = [2, 3]

W = [4, 5]

b = 1

The neuron calculates:

z = (2 × 4) + (3 × 5) + 1

z = 8 + 15 + 1

z = 24

Then:

a = ReLU(24)

Since 24 is positive:

a = 24

```python
x = np.array([2, 3])
W = np.array([4, 5])
b = 1

z = np.dot(x, W) + b

print("z =", z)
```

```python
a = max(0, z)

print("a =", a)
```

## 3. ReLU Activation Function

ReLU stands for:

Rectified Linear Unit

Its rule is:

ReLU(z) = z       if z > 0
ReLU(z) = 0       if z ≤ 0

### ELI5

Think of ReLU as a gate.

Positive number:
→ Gate opens
→ Number passes through

Negative number:
→ Gate closes
→ Output becomes 0

Examples:

z = 7  → ReLU = 7

z = 2  → ReLU = 2

z = -4 → ReLU = 0

z = 0  → ReLU = 0

```python
def relu(z):
    return np.maximum(0, z)

values = np.array([7, 2, -4, 0])

print("Input:", values)
print("ReLU:", relu(values))
```

## 4. ReLU Derivative

For backpropagation, we need the derivative of ReLU.

The simple rule is:

da/dz = 1    if z > 0

da/dz = 0    if z < 0

For our examples, we treat z = 0 as having derivative 0.

### ELI5

ReLU is like a gate.

If z is positive:

Gate = OPEN

Derivative = 1

If z is negative:

Gate = CLOSED

Derivative = 0

So:

Positive → gradient passes

Negative → gradient is blocked

```python
def relu_derivative(z):
    return 1 if z > 0 else 0

print("Derivative at z = 5:", relu_derivative(5))
print("Derivative at z = -5:", relu_derivative(-5))
print("Derivative at z = 0:", relu_derivative(0))
```

## 5. Chain Rule

The loss does not directly depend on z.

The path is:

z → a → Loss

Therefore we use the chain rule:

dLoss/dz = dLoss/da × da/dz

### ELI5

Imagine:

z
 ↓
ReLU
 ↓
a
 ↓
Loss

To understand how changing z affects Loss,
we follow the path backward.

We multiply the effects at each step.

## 6. Chain Rule Example

Suppose:

dLoss/da = 6

and:

a = z

Because:

da/dz = 1

Therefore:

dLoss/dz

= dLoss/da × da/dz

= 6 × 1

= 6

```python
dLoss_da = 6
da_dz = 1

dLoss_dz = dLoss_da * da_dz

print("dLoss/dz =", dLoss_dz)
```

## 7. ReLU + Chain Rule

Suppose:

z = 5

dLoss/da = 4

Because z is positive:

da/dz = 1

Therefore:

dLoss/dz

= 4 × 1

= 4

The gradient passes through.

```python
z = 5
dLoss_da = 4

da_dz = relu_derivative(z)

dLoss_dz = dLoss_da * da_dz

print("da/dz =", da_dz)
print("dLoss/dz =", dLoss_dz)
```

## 8. ReLU Can Block Gradients

Suppose:

z = -5

dLoss/da = 4

Because z is negative:

da/dz = 0

Therefore:

dLoss/dz

= 4 × 0

= 0

The gradient is blocked.

### ELI5

The gradient reaches the ReLU gate.

The gate sees:

z < 0

So the gate is closed.

Nothing passes backward.

Therefore:

dLoss/dz = 0

```python
z = -5
dLoss_da = 4

da_dz = relu_derivative(z)

dLoss_dz = dLoss_da * da_dz

print("da/dz =", da_dz)
print("dLoss/dz =", dLoss_dz)
```

## 9. Weight Gradient

We already learned this in Day 13.

For a neuron:

z = W · x + b

The gradient for each weight is:

dLoss/dW = dLoss/dz × x

For example:

x = [2, 3]

dLoss/dz = 4

Therefore:

dLoss/dW

= 4 × [2, 3]

= [8, 12]

```python
x = np.array([2, 3])
dLoss_dz = 4

dLoss_dW = dLoss_dz * x

print("dLoss/dW =", dLoss_dW)
```

## 10. Bias Gradient

Our neuron is:

z = W · x + b

The derivative of z with respect to b is:

dz/db = 1

Therefore:

dLoss/db

= dLoss/dz × dz/db

= dLoss/dz × 1

Therefore:

dLoss/db = dLoss/dz

### ELI5

Bias is directly added to z.

If bias increases by 1:

z also increases by 1.

So:

dz/db = 1

```python
dLoss_dz = 2

dLoss_db = dLoss_dz

print("dLoss/db =", dLoss_db)
```

## 11. Complete Single-Neuron Example

Given:

x = [3, 2]

W = [4, 5]

b = -2

dLoss/da = 3

### Forward Pass

z = (3 × 4) + (2 × 5) - 2

z = 12 + 10 - 2

z = 20

Because z > 0:

a = ReLU(20)

a = 20

### Backward Pass

Since z > 0:

da/dz = 1

Therefore:

dLoss/dz

= dLoss/da × da/dz

= 3 × 1

= 3

Weight gradient:

dLoss/dW

= dLoss/dz × x

= 3 × [3, 2]

= [9, 6]

Bias gradient:

dLoss/db = dLoss/dz

= 3

Final gradients:

dW = [9, 6]

db = 3

```python
x = np.array([3, 2])
W = np.array([4, 5])
b = -2

dLoss_da = 3

# Forward pass
z = np.dot(x, W) + b
a = relu(z)

# Backward pass
da_dz = relu_derivative(z)

dLoss_dz = dLoss_da * da_dz
dLoss_dW = dLoss_dz * x
dLoss_db = dLoss_dz

print("z =", z)
print("a =", a)
print("da/dz =", da_dz)
print("dLoss/dz =", dLoss_dz)
print("dLoss/dW =", dLoss_dW)
print("dLoss/db =", dLoss_db)
```

## 12. Multiple Neurons

Now let's extend the same idea to multiple neurons.

Input:

x = [2, 3]

Weight matrix:

W = [
    [4, 5],
    [1, 2]
]

Each row represents one neuron.

Neuron 1:

[4, 5]

Neuron 2:

[1, 2]

Forward pass:

z = W @ x

Neuron 1:

(4 × 2) + (5 × 3)
= 23

Neuron 2:

(1 × 2) + (2 × 3)
= 8

Therefore:

z = [23, 8]

```python
x = np.array([2, 3])

W = np.array([
    [4, 5],
    [1, 2]
])

z = W @ x

print("z =", z)
```

## 13. Gradients for Multiple Neurons

Suppose:

dLoss/dz = [4, -3]

Input:

x = [2, 3]

For Neuron 1:

dLoss/dz1 = 4

Gradient:

4 × [2, 3]

= [8, 12]

For Neuron 2:

dLoss/dz2 = -3

Gradient:

-3 × [2, 3]

= [-6, -9]

Therefore the complete gradient matrix is:

dLoss/dW = [
    [ 8, 12],
    [-6, -9]
]

Notice:

The gradient matrix has the same shape as the weight matrix.

```python
x = np.array([2, 3])

dLoss_dz = np.array([4, -3])

dLoss_dW = dLoss_dz[:, None] * x

print("Gradient matrix:")
print(dLoss_dW)
```

## 14. Updating Multiple Weights

Gradient descent formula:

W_new = W_old - learning_rate × gradient

Suppose:

W = [
    [4, 5],
    [1, 2]
]

Gradient = [
    [8, 12],
    [-6, -9]
]

Learning rate = 0.1

The same formula is applied to every weight.

Positive gradient:

→ weight decreases

Negative gradient:

→ weight increases

```python
W = np.array([
    [4.0, 5.0],
    [1.0, 2.0]
])

gradient = np.array([
    [8.0, 12.0],
    [-6.0, -9.0]
])

learning_rate = 0.1

new_W = W - learning_rate * gradient

print("Old W:")
print(W)

print("\nGradient:")
print(gradient)

print("\nNew W:")
print(new_W)
```

And:

### The key formulas

1. Forward:

3. Chain rule:

Backpropagation is essentially:

1. Find how Loss changes with the output.
2. Move backward through each operation.
3. Multiply the derivatives using the chain rule.
4. Calculate gradients for weights and biases.
5. Update the parameters using gradient descent.
