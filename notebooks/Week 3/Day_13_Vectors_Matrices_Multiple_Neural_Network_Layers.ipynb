## 1. Vectors

A vector is a collection of numbers.

Example:

x = [2, 4, 3]

In a neural network, this can represent three input features.

We can think of:

- x₁ = 2
- x₂ = 4
- x₃ = 3

```python
x = np.array([2, 4, 3])

print("x: ", x)
print("shape of x: ", x.shape)
```

## 2. Dot Product

For one neuron:

z = x · W + b

The dot product means:

1. Multiply corresponding values.
2. Add all the results.
3. Add the bias.

Example:

x = [2, 4, 3]

W = [5, 2, 1]

b = 2

Therefore:

z = (2×5) + (4×2) + (3×1) + 2

z = 10 + 8 + 3 + 2

z = 23

```python
x = np.array([2, 4, 3])
W = np.array([5, 2, 1])
b = 2

z = np.dot(x, W) + b

print("z = ", z)
```

## 3. Multiple Neurons

Multiple neurons can receive the same input.

However, each neuron has its own weights and bias.

Input:

x = [2, 4, 3]

Neuron 1:

W1 = [5, 2, 1]
b1 = 2

Neuron 2:

W2 = [1, 3, 4]
b2 = 1

Each neuron produces one output.

```python
x = np.array([2, 4, 3])

W1 = np.array([5, 2, 1])
W2 = np.array([1, 3, 4])

b1 = 2
b2 = 1

z1 = np.dot(x, W1) + b1
z2 = np.dot(x, W2) + b2

print("neuron 1 output ", z1)
print("neuron 2 output ", z2)

print("Output vector = ", np.array([z1, z2]))
```

## 4. Weight Matrix

Instead of keeping each neuron's weights separately, we can put them into a matrix.

W = [
    [5, 2, 1],   ← Neuron 1
    [1, 3, 4]    ← Neuron 2
]

### Important rule

For our convention:

- Rows = neurons
- Columns = inputs
- Number of rows = number of outputs

Therefore:

A 2 × 3 matrix means:

- 2 neurons
- 3 inputs per neuron
- 2 outputs

```python
W = np.array([
    [5, 2, 1],
    [1, 3, 4]
])

print("Weight matrix:")
print(W)

print("Shape:", W.shape)
```

## 5. Matrix Multiplication

Each row of the weight matrix is dotted with the input vector.

W = [
    [5, 2, 1],
    [1, 3, 4]
]

x = [2, 4, 3]

Neuron 1:

[5, 2, 1] · [2, 4, 3]

= 10 + 8 + 3

= 21

Neuron 2:

[1, 3, 4] · [2, 4, 3]

= 2 + 12 + 12

= 26

Therefore:

W @ x = [21, 26]

```python
x = np.array([2, 4, 3])

W = np.array([
    [5, 2, 1],
    [1, 3, 4]
])

z = W @ x

print("W @ x =", z)
```

## 6. Bias Vector

Each neuron has its own bias.

Suppose:

W @ x = [21, 26]

and:

b = [2, 1]

Then:

z = [21, 26] + [2, 1]

z = [23, 27]

So the final output of the layer is:

[23, 27]

```python
x = np.array([2, 4, 3])

W = np.array([
    [5, 2, 1],
    [1, 3, 4]
])

b = np.array([2, 1])

z = W @ x + b

print("W @ x:", W @ x)
print("Bias:", b)
print("Final output:", z)
```

## 7. A 3 × 3 Weight Matrix

Consider:

W = [
    [2, 1, 3],
    [4, 2, 1],
    [1, 5, 2]
]

x = [1, 2, 3]

A 3 × 3 matrix means:

- 3 neurons
- 3 inputs per neuron
- 3 outputs

```python
x = np.array([1, 2, 3])

W = np.array([
    [2, 1, 3],
    [4, 2, 1],
    [1, 5, 2]
])

print("W @ x =", W @ x)
```

```python
b = np.array([1, 2, 3])

z = W @ x + b

print("W @ x:", W @ x)
print("Bias:", b)
print("Final output:", z)
```

## 8. Multiple Neural Network Layers

The output of one layer becomes the input to the next layer.

Example:

Input:

[1, 2, 3]

        ↓

Layer 1

[14, 13, 20]

        ↓

Layer 2

[101, 120]

The important idea is:

> The number of inputs to a layer equals the number of outputs from the previous layer.

```python
# Input

x = np.array([1, 2, 3])

# Layer 1
# 3 neurons, 3 inputs per neuron

W1 = np.array([
    [2, 1, 3],
    [4, 2, 1],
    [1, 5, 2]
])

b1 = np.array([1, 2, 3])

layer1_output = W1 @ x + b1

print("Layer 1 output:", layer1_output)

# Layer 2
# 2 neurons, 3 inputs per neuron

W2 = np.array([
    [2, 1, 3],
    [1, 2, 4]
])

b2 = np.array([0, 0])

layer2_output = W2 @ layer1_output + b2

print("Layer 2 output:", layer2_output)
```

## 9. Understanding Matrix Shapes

For our convention:

Weight matrix shape:

current layer neurons × previous layer outputs

Example:

Input = 3 values

Layer 1 = 5 neurons

Layer 2 = 4 neurons

Therefore:

W1 = 5 × 3

W2 = 4 × 5

Why?

W1:

5 neurons × 3 inputs

W2:

4 neurons × 5 inputs

The 5 inputs to Layer 2 come from the 5 outputs of Layer 1.

```python
# Example shapes

W1 = np.zeros((5, 3))
W2 = np.zeros((4, 5))

print("W1 shape:", W1.shape)
print("W2 shape:", W2.shape)
```

## 10. Gradient for Multiple Weights

For one neuron, we learned:

dLoss/dW = dLoss/dz × dz/dW

Because:

dz/dW = x

we get:

dLoss/dW = dLoss/dz × x

Example:

x = [3, 2, 4]

dLoss/dz = -2

Therefore:

dLoss/dW1 = -2 × 3 = -6

dLoss/dW2 = -2 × 2 = -4

dLoss/dW3 = -2 × 4 = -8

Gradient = [-6, -4, -8]

```python
x = np.array([3, 2, 4])

dLoss_dz = -2

gradient = dLoss_dz * x

print("Gradient:", gradient)
```

## 11. Weight Update

Gradient descent updates each weight using:

new_weight = old_weight - learning_rate × gradient

Example:

W = [4, 2, 1]

Gradient = [12, 8, 20]

Learning rate = 0.1

W1:

4 - (0.1 × 12)
= 2.8

W2:

2 - (0.1 × 8)
= 1.2

W3:

1 - (0.1 × 20)
= -1

Therefore:

New W = [2.8, 1.2, -1]

```python
W = np.array([4.0, 2.0, 1.0])

gradient = np.array([12.0, 8.0, 20.0])

learning_rate = 0.1

new_W = W - learning_rate * gradient

print("Old W:", W)
print("Gradient:", gradient)
print("New W:", new_W)
```

```python
import numpy as np

# Input
x = np.array([1, 2, 3])

# Layer 1: 3 neurons
W1 = np.array([
    [2, 1, 3],
    [4, 2, 1],
    [1, 5, 2]
])

b1 = np.array([1, 2, 3])

# Forward pass through Layer 1
z1 = W1 @ x + b1

# Layer 2: 2 neurons
W2 = np.array([
    [2, 1, 3],
    [1, 2, 4]
])

b2 = np.array([0, 0])

# Forward pass through Layer 2
z2 = W2 @ z1 + b2

print("Input:", x)
print("Layer 1 output:", z1)
print("Layer 2 output:", z2)
```
