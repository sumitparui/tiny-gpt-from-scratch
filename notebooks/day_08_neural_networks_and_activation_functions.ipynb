Today we learn why neural networks also use activation functions.

A neural-network layer can be written as:

output = input @ weights + bias

An activation function is then applied to the output:

output = activation(input @ weights + bias)

Activation functions introduce non-linearity.

This allows neural networks to learn more complicated patterns.

## ReLU

ReLU stands for Rectified Linear Unit.

Its rule is:

ReLU(x) = max(0, x)

Therefore:

negative number → 0
positive number → unchanged

Example:

[-3, 2, -1, 5]

After ReLU:

[0, 2, 0, 5]

## A Complete Neural-Network Layer

A neural-network layer performs two main steps.

### Step 1 — Linear transformation

z = input @ weights + bias

### Step 2 — Activation

output = ReLU(z)

Therefore:

output = ReLU(input @ weights + bias)

Example:

Input:

[2, 1]

Weights:

[
 [1, -2],
 [3, -1]
]

Bias:

[0, -2]

First:

[2, 1] @
[
 [1, -2],
 [3, -1]
]

= [5, -5]

Add bias:

[5, -5] + [0, -2]

= [5, -7]

Apply ReLU:

[5, -7] → [5, 0]

Final output:

[5, 0]

## Multiple Neural-Network Layers

A neural network can stack multiple layers.

Example:

Input
↓
Layer 1
↓
ReLU
↓
Layer 2
↓
ReLU
↓
Output

For our example:

Input:

[2, 1]

Layer 1:

[2, 1] @
[
 [1, -2],
 [3, -1]
]

= [5, -5]

Add bias:

[5, -5] + [0, -2]

= [5, -7]

Apply ReLU:

[5, -7] → [5, 0]

Therefore Layer 1 output:

[5, 0]

This becomes the input to Layer 2.

Layer 2:

[5, 0] @
[
 [2, 1],
 [3, 4]
]

= [10, 5]

Add bias:

[10, 5] + [1, 2]

= [11, 7]

Apply ReLU:

[11, 7] → [11, 7]

Final network output:

[11, 7]

The important idea is:

The output of one layer becomes the input
to the next layer.

```python
import numpy as np

input_vector = np.array([1, 2])

weights = np.array([
    [2, 3],
    [4, 5]
])

bias = np.array([1, 1])

z = (input_vector @ weights) + bias

#Linear Activation
output = np.maximum(0, z)

print("Input:", input_vector)
print("Weights:")
print(weights)
print("Bias:", bias)
print("After linear transformation:", z)
print("After ReLU:", output)
```
