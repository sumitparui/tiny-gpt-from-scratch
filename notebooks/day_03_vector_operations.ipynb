```python
import numpy as np

A = np.array([2, 4, 6])
B = np.array([1, 3, 5])

print("A + B", A + B)
print("A - B", A - B)
print("2 x A", 2 * A)
```

Add The Magnitude example

```python
v = np.array([3, 4])

magnitude = np.sqrt(np.sum(v ** 2))

print("vector: ", v)
print("magnitude ", magnitude)
```
