# Day 9 — Building a Neural Network From Scratch

Today we will implement a simple neural-network layer
using Python and NumPy.

The layer performs:

1. Matrix multiplication
2. Add bias
3. Apply ReLU

The mathematical formula is:

z = input @ weights + bias

output = ReLU(z)

```python
import numpy as np

x = np.array([-3, 2, -1, 5])

result = np.maximum(0, x)

print(result)
```

```python
def neural_layer(input_vector, weights, bias):
  z = (input_vector @ weights) + bias
  output = np.maximum(0, z)
  return output
```

```python
input_vector=np.array([1, 2])

weights = np.array([
    [2, 3],
    [4, 5]
])

bias = np.array([1, 1])

# z = (input_vector @ weights) + bias

# output = np.maximum(0, z)

output = neural_layer(input_vector, weights, bias)
# print("Linear output:", z)
print("Output :", output)
```

1. Matrix multiplication >
input @ weights
2. Bias >
input @ weights + bias
3. ReLU >
ReLU(x) = max(0, x)
4. Complete layer >
output = ReLU(input @ weights + bias)
5. Multiple layers >
Layer 1 output → Layer 2 input
