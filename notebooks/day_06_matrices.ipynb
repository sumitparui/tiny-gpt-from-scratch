# Day 6 — Matrices

A matrix is a rectangular arrangement of numbers.

Example:

[
    [0.2, 0.4, 0.6],
    [0.1, 0.3, 0.5],
    [0.7, 0.8, 0.9]
]

This matrix has:
- 3 rows
- 3 columns

Therefore its shape is 3 × 3.

A vector is a single list of numbers.
A matrix can contain multiple vectors arranged as rows or columns.

```python
import numpy as np

matrix = np.array([
     [0.2, 0.4, 0.6],
    [0.1, 0.3, 0.5],
    [0.7, 0.8, 0.9]

])

print(matrix)
```

## Rows and Columns

For the matrix:

[
    [0.2, 0.4, 0.6],
    [0.1, 0.3, 0.5],
    [0.7, 0.8, 0.9]
]

Rows:
- Row 1 = [0.2, 0.4, 0.6]
- Row 2 = [0.1, 0.3, 0.5]
- Row 3 = [0.7, 0.8, 0.9]

Columns:
- Column 1 = [0.2, 0.1, 0.7]
- Column 2 = [0.4, 0.3, 0.8]
- Column 3 = [0.6, 0.5, 0.9]

```python
print("Matrix shape", matrix.shape)
```

## Embeddings and Matrices

Each token can have an embedding vector.

Example:

cat → [0.2, 0.4, 0.6]
dog → [0.1, 0.3, 0.5]
car → [0.7, 0.8, 0.9]

When multiple embedding vectors are arranged together,
they form a matrix.

Therefore:

Multiple vectors → Matrix

## Sentence as a Matrix

Suppose a sentence has 3 tokens and each token
has a 4-dimensional embedding.

Then the embedding matrix has shape:

3 × 4

because:

3 = number of tokens
4 = embedding dimensions

So:

Number of tokens × Embedding dimension
= Embedding matrix shape

```python
print("First row: ", matrix[0])
print("Second row: ", matrix[1])
print("Third row: ", matrix[2])
```

## Accessing Matrix Rows and Columns

NumPy uses:

matrix[rows, columns]

The `:` means "all".

Therefore:

matrix[:, 0]

means:

- all rows
- column 0

Example:

[
 [0.2, 0.4, 0.6],
 [0.1, 0.3, 0.5],
 [0.7, 0.8, 0.9]
]

matrix[:, 0]

returns:

[0.2, 0.1, 0.7]

```python
print("First columns", matrix[:, 0])
print("Second columns", matrix[:, 1])
```

## Matrix Addition

Matrix addition works by adding corresponding elements.

Example:

A =
[
 [1, 2],
 [3, 4]
]

B =
[
 [5, 6],
 [7, 8]
]

A + B =

[
 [1+5, 2+6],
 [3+7, 4+8]
]

=
[
 [6, 8],
 [10, 12]
]

For basic matrix addition, both matrices must have the same shape.

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

C = A + B

print("A: ")
print(A)

print("\nB: ")
print(B)

print("\nC: ")
print(C)

```

## Scalar Multiplication

A scalar is a single number.

Scalar multiplication means multiplying every element
of a matrix by the scalar.

Example:

A =
[
 [1, 2],
 [3, 4]
]

3 × A =

[
 [3, 6],
 [9, 12]
]

The shape does not change.

A is 2 × 2.
3 × A is also 2 × 2.

```python
A = np.array([
    [1, 2],
    [3, 4]
])

result = 3 * A

print("A: ")
print(A)

print("\nresult: ")
print(result)

print("\nshape: ", result.shape)
```

# Matrix Multiplication

For matrix multiplication:

A = m × n
B = n × p

The inside dimensions must match:

(m × n) × (n × p)

The resulting matrix has the outside dimensions:

m × p

Example:

A = 2 × 2
B = 2 × 2

Therefore:

A × B = 2 × 2

## How Matrix Multiplication Works

Matrix multiplication uses dot products.

For:

A =
[
 [1, 2],
 [3, 4]
]

B =
[
 [5, 6],
 [7, 8]
]

The top-left result:

[1, 2] · [5, 7]
= (1×5) + (2×7)
= 19

The top-right result:

[1, 2] · [6, 8]
= (1×6) + (2×8)
= 22

The bottom-left result:

[3, 4] · [5, 7]
= 43

The bottom-right result:

[3, 4] · [6, 8]
= 50

Therefore:

A × B =
[
 [19, 22],
 [43, 50]
]

Matrix multiplication can be understood as multiple
dot products between rows of the first matrix and
columns of the second matrix.

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

result = A @ B

print("A: ")
print(A)

print("\nB: ")
print(B)

print("\nA x B: ")
print(result)

```

## Matrix Multiplication vs Element-wise Multiplication

These are different operations.

Element-wise multiplication:

A * B

multiplies corresponding elements.

Matrix multiplication:

A @ B

uses rows of A and columns of B and calculates
dot products.

Example:

A @ B

means matrix multiplication.

A * B

means element-wise multiplication in NumPy.

```python
A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([
    [6, 7],
    [8, 9]
])

print("Element wise multiplication: ")
print(A * B)

print("\nmatrix multiplication: ")
print(A @ B)
```

## Why Matrix Multiplication Matters in AI

Neural networks represent information using numbers,
vectors, and matrices.

A neural network contains weight matrices.

A simplified view is:

Input
↓
Matrix multiplication with weights
↓
New values
↓
More calculations

Matrix multiplication allows neural networks to
transform numerical representations.

This operation appears throughout neural networks
and Transformer models.
