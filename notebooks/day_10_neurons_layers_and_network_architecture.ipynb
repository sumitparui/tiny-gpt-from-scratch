## What is a Neuron?

A neuron is a small mathematical unit that takes inputs,
multiplies them by weights, adds a bias, and applies an
activation function.

The basic calculation is:

z = input · weights + bias

Then:

output = activation(z)

For our examples, we use ReLU as the activation function.

ReLU(x) = max(0, x)

## One Neuron Example

Suppose:

Input = [2, 3]

Weights = [4, 5]

Bias = 2

The neuron calculates:

z = (2 × 4) + (3 × 5) + 2

z = 8 + 15 + 2

z = 25

Then:

ReLU(25) = 25

Therefore, the neuron produces one output:

25

```python
import numpy as np
```

```python
input_vector = np.array([2, 3])

weights = np.array([4, 5])

bias = 2

z = input_vector @ weights + bias

output = np.maximum(0, z)

print("Output:", output)
```

## Two Neurons

A single neuron produces one output.

If we have two neurons, each neuron has its own weights and bias.

Input:

[2, 3]

Neuron 1:

Weights = [4, 5]
Bias = 2

Output:

(2 × 4) + (3 × 5) + 2
= 25

Neuron 2:

Weights = [1, 2]
Bias = 1

Output:

(2 × 1) + (3 × 2) + 1
= 9

Therefore:

Neuron 1 → 25
Neuron 2 → 9

Layer output:

[25, 9]

## Representing Multiple Neurons With a Weight Matrix

Instead of calculating each neuron separately,
we can represent their weights using a matrix.

Input:

[2, 3]

Neuron 1 weights:

[4, 5]

Neuron 2 weights:

[1, 2]

Weight matrix:

[
 [4, 1],
 [5, 2]
]

The columns represent the neurons.

Column 1 → Neuron 1 → [4, 5]

Column 2 → Neuron 2 → [1, 2]

Bias:

[2, 1]

```python
input_vector = np.array([2, 3])

weights = np.array([
    [4, 1],
    [5, 2]
])

bias = np.array([2, 1])

output = input_vector @ weights + bias

output = np.maximum(0, output)

print("Output:", output)
```

## What is a Layer?

A layer is a collection of neurons that process the same input.

For example:

Input
  ↓
Neuron 1 → Output 1
Neuron 2 → Output 2
Neuron 3 → Output 3

If a layer contains 3 neurons,
the layer produces 3 output values.

Therefore:

1 neuron → 1 output
2 neurons → 2 outputs
100 neurons → 100 outputs

## Complete Layer Calculation

A neural-network layer can be written as:

z = input @ weights + bias

Then apply the activation function:

output = ReLU(z)

Therefore:

output = ReLU(input @ weights + bias)

The process is:

Input
↓
Matrix multiplication
↓
Add bias
↓
ReLU
↓
Layer output

```python
input_vector = np.array([2, 3])

weights = np.array([
    [4, 2],
    [1, 5]
])

bias = np.array([2, 1])

z = input_vector @ weights + bias

output = np.maximum(0, z)

print("Linear output:", z)
print("Layer output:", output)
```

## Negative Values and ReLU

ReLU removes negative values.

Example:

Input:

[2, 1]

Weights:

[
 [1,  3],
 [2, -4]
]

Bias:

[-5, 1]

The calculations produce:

Neuron 1:

Weighted sum = 4
Add bias = -1
ReLU = 0

Neuron 2:

Weighted sum = 2
Add bias = 3
ReLU = 3

Therefore:

Layer output = [0, 3]

```python
input_vector = np.array([2, 1])

weights = np.array([
    [1, 3],
    [2, -4]
])

bias = np.array([-5, 1])

z = input_vector @ weights + bias

output = np.maximum(0, z)

print("Linear output:", z)
print("Layer output:", output)
```

## Number of Weights and Biases

Suppose a layer has:

4 inputs
3 neurons

Each neuron needs one weight for every input.

Therefore:

Neuron 1 → 4 weights
Neuron 2 → 4 weights
Neuron 3 → 4 weights

Total weights:

4 × 3 = 12

Each neuron has one bias.

Therefore:

3 neurons → 3 biases

So:

Weights = 12
Biases = 3

The weight matrix has shape:

(4, 3)

The bias vector has shape:

(3,)

## Important Rules

If a layer has:

N inputs
M neurons

Then:

Number of weights = N × M

Number of biases = M

The output contains M values.

Therefore:

4 inputs + 3 neurons

→ 12 weights
→ 3 biases
→ 3 outputs
