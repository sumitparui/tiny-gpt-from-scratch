```python
import numpy as np
from src.embeddings import get_embedding

embedding_matrix = np.array([
     [0.1, 0.2, 0.3],   # I
    [0.4, 0.5, 0.6],   # love
    [0.7, 0.8, 0.9],   # cat
    [1.0, 1.1, 1.2]    # AI
])

embedding = get_embedding(2, embedding_matrix)
print(embedding)
```

```python
vocab={
    "I" : 0,
    "love": 1,
    "cat" : 2,
    "AI" : 3
}

text = "I love cat"

tokens = text.split()
token_ids = [vocab[i] for i in tokens]

print("Tokens: ", tokens)
print("Token Id:", token_ids)

```

```python
embeddings = [embedding_matrix[token_id]  for token_id in token_ids]
print("embedding: ", embeddings)
```

```python
text = "I love AI"

tokens = text.split()

token_ids = [vocab[token] for token in tokens]

embeddings = [embedding_matrix[token_id] for token_id in token_ids]

print("Text:", text)
print("Tokens:", tokens)
print("Token IDs:", token_ids)
print("Embeddings:", embeddings)
```
